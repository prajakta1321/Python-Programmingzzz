```python
def double(x):
  print("doubling {x}")
  return x*2

  numbers = [1,2,3,4,5]
  doubled_numbers = list(map(double, numbers))

  print(doubled_numbers)
```

```python
for number in doubled_numbers:
  print(number)
print("map function completed")
```

```python
# eager evaluation using list comprehension

num = [1,2,3,4,5]
square = [n**2 for n in num]

for s in square:
  print(s)
```

```python
# lazy evaluation using map()

num = [1,2,3,4,5]
sq = map(lambda x : x**2, num)

for s in sq:
  print(s)
```

```python
# using an iterator
my_iterator = iter([1,2,3,4,5])
for i in my_iterator:
  print(i)
```

```python
# using a list
my_list = [1,2,3,4,5]
for i in my_list:
  print(i)
```

```python
# accessing a list multiple times
my_list = [1,2,3,4,5]
print(my_list)
```

```python
def my_range(start, end):
  current = start

  while current < end:
    yield current
    current += 1
my_iterator = my_range(0,5)
print(next(my_iterator))
```

```python
numbers = [1, 2, 3, 4, 5]
doubled_numbers = map(lambda x: x * 2, numbers)
print(list(doubled_numbers)) # Output : [2, 4, 6, 8, 10]
```

```python
def square(x):
  return x ** 2
numbers = [1, 2, 3, 4, 5]
squares = map(square, numbers)
print(list(squares)) # Output: [1, 4, 9, 16, 25]
```

```python
numbers = [1, 2, 3, 4, 5]
squares = map(lambda x: x ** 2, numbers)
print(list(squares)) # Output: [1, 4, 9, 16, 25]
```
